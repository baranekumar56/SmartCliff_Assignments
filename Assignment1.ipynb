3. Write a program to print all prime numbers between 1 and N(inclusive). N will be
the input.


```python
from math import sqrt

# for finding the whether a

def is_prime(n):
    if n < 2:
        return False

    if n == 2:
        return True

    for i in range(2, int(sqrt(n) + 1)):
        if n % i == 0:
            return False

    return True

n = int(input())
print("Input :", n)
for i in range(n+1):
    if is_prime(i):
        print(i)
```

Above 300 ₹ 6.0

First 100 units at ₹1.5
Next 100 units at ₹2.5
Remaining 50 units at ₹4.0

Also, if the input provided is not exactly the factorial of a number, say, the input
provided is 122, which is not a perfect factorial of a number, it should return “Sorry.
The given number is not a perfect factorial”.


```python


def is_factorial_number(n):

    if n <= 0:
        print("Invalid input")
        return

    fact = 1
    nums = 1

    while fact <= n:
        if n == fact:
            print(nums)
            return

        nums += 1

        fact *= nums

    print("Sorry. The given number is not a perfect factorial")



n = int(input())
print("Input :", n)
is_factorial_number(n)

```

16. Divisor Game
Problem statement: Alice and Bob take turns playing a game, with Alice starting first.
Initially, there is a number n on the chalkboard. On each player's turn, that player makes a
move consisting of:
● Choosing any x with 0 < x < n and n % x == 0.
● Replacing the number n on the chalkboard with n - x.
 Also, if a player cannot make a move, they lose the game.
 Return true only if Alice wins the game, assuming both players play optimally

```python
# turn 0 represents Alice and turn 1 represents Bob

dp = [[-1, -1]] * 1001

def get_factor(n):
    factors = [1]

    for i in range(2, n):
        if n % i == 0:
            factors.append(i)

    return factors

def f(n, turn):

    if n <= 1 and turn == 0:
        return False

    if n <= 1 and turn == 1:
        return True

    if dp[n][turn] != -1:

        return dp[n][turn]

    factors = get_factor(n)
    # the current turn player have the ability to choose from any of these factors
    res = False
    for i in factors:
        if f(n - i, 1 if turn == 0 else 0):
            res = True


    dp[n][turn] = res
    return dp[n][turn]


n = int(input())
print("Input :", n)
print(f(n, 0))

```

19. Climbing Stairs
Problem Statement: You are climbing a staircase. It takes n steps to reach the top. Each time
you can either climb 1 or 2 steps. In how many distinct ways can you climb to the top?


```python
n = int(input())
print("Input :", n)

# i have to find the nth fibonacci number

def get_nth_fib(n):
    if n <= 2:
        return n

    a = 1
    b = 1
    c = a + b

    for i in range(3, n + 1):
        a = b
        b = c
        c = a + b

    return c


print(get_nth_fib(n))
```

22. Collatz Sequence Steps
Given a positive integer n, repeatedly apply the following rules:
• If n is even, divide it by 2.
• If n is odd, multiply it by 3 and add 1.
Count how many steps it takes to reach 1.

```python
n = int(input())
print("Input :", n)

def calculate_steps(n):

    if n == 1: return 0

    if n % 2 == 0:
        return 1 + calculate_steps(n // 2)

    else:
        return 1 + calculate_steps(3 * n + 1)

calculate_steps(n)
```

24. Josephus Problem
Given n people standing in a circle and eliminating every kth person until one remains, find the
position of the survivor (1-indexed).


```python
n, k = list(map(int, input().split(',')))

# initialize the list with n numbers, delete the K th element, increment k by k and assign k = k + k % length of list
# repeat the process until one element is remaining

print("Input :", n, k)

def josephus(n, k):

    k -= 1

    i = len(n) - 1
    ind = k
    while i :

        n.pop(ind)

        ind += k
        ind = ind % len(n)

        i -= 1

    return n

n = [i for i in range(1, n + 1)]

josephus(n, k)
```

27. Trap Rainwater Problem
Given heights of blocks, calculate how much water can be trapped between them after rainfall.

```python
nums = list(map(int, input().split()))
print("Input :", nums)

# we can set two pointers l and r, and two variables for stroing right max and left max value,
# on each iteration update lmax and rmax ,
# when lmax greater then rmax move r pointer , else move l pointer, on the same time calculate the water
# storable at that location for each iteration

def trapping_rain_water(nums):

    l = 1
    r = len(nums) - 2

    lmax = nums[0]
    rmax = nums[-1]

    if len(nums) <= 2:
        return 0

    storable_water = 0

    while l <= r:
        lmax = max(lmax, nums[l])
        rmax = max(rmax, nums[r])

        if lmax > rmax:
            storable_water += rmax - nums[r]
            r -= 1

        else:
            storable_water += lmax - nums[l]
            l += 1

    return storable_water

trapping_rain_water(nums)



```

28. Circular Prime Check
A number is circular prime if all rotations of its digits are prime. Check if a given number
is a circular prime.

```python
from math import sqrt
n  = input()

print("Input :", n)

def is_prime(n):
    n = int(n)

    if n <= 1: return False

    if n == 2: return True

    for i in range(2, int(sqrt(n)+1)):
        if n % i == 0: return False
    return True

not_circular_prime = False

s = str(n)

for i in range(0, len(s)):
    if not is_prime(n):
        not_circular_prime = True
        break

    n = list(n)
    n.append(n[0])
    n.pop(0)
    n = "".join(n)

if not_circular_prime:
    print("NO")
else :
    print("YES")
```

29. Kaprekar’s Process
Given a 4-digit number with at least two different digits, repeatedly:
• Arrange digits in descending and ascending order to form two numbers.
• Subtract the smaller from the larger.
• Repeat until you reach 6174.
Count the steps required.

```python
n = input()

# the main approach for this problem is, we get the input from the user then
# sort it in ascending and descending order then convert it to integer and check if the result is 6174
# if it is , print the count , else continue

steps = 0

while n != '6174':
    a = list(n)
    b = list(n)

    a = "".join(sorted(a))
    b = "".join(sorted(b, reverse=True))
    print(a, b)
    steps += 1

    n = str(abs(int(a) - int(b))).zfill(4)
    print(n)


print(steps)
```

30. Digital Root Multiplication
